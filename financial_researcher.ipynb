**Step 6: Run the code to use imported project from Github**

```python
import os
import sys
from datetime import datetime

import gradio as gr
from google.colab import userdata

# API Keys
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
os.environ["SERPER_API_KEY"] = userdata.get("SERPER_API_KEY")

# Project Path
PROJECT_ROOT = "/content/AgenticAI-Lab/financial_researcher"
sys.path.insert(0, f"{PROJECT_ROOT}/src")

from financial_researcher.crew import FinancialResearcher


def research_company(company):

    if not company.strip():
        return "Please enter a company name."

    try:
        inputs = {
            "company": company,
            "current_date": str(datetime.now().date())
        }

        result = FinancialResearcher().crew().kickoff(inputs=inputs)

        return str(result)

    except Exception as e:
        return f"Error:\n\n{e}"
```

**Step 7: Run the code with gradio UI**

```python
demo = gr.Interface(
    fn=research_company,
    inputs=gr.Textbox(
        label="Company Name",
        placeholder="e.g. Apple, Microsoft, Nvidia"
    ),
    outputs=gr.Textbox(
        label="Research Report",
        lines=30
    ),
    title="📈 Financial Research Agent",
    description="""
Enter a public company name.

The CrewAI Financial Researcher will:
• Search recent news
• Analyze financial information
• Generate an investment research report
"""

)

demo.launch(share=True)
```
